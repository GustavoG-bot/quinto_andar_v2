### Extração da tabela

```python
import os

# Configurando a JVM manualmente
os.environ['JAVA_HOME'] = r"C:\Program Files\Java\jre1.8.0_441"
os.environ['PATH'] += os.pathsep + os.path.join(os.environ['JAVA_HOME'], 'bin')
```

```python
import tabula
import pandas as pd

arquivo_pdf = "ranking-abras-2024-completo.pdf"
tabelas = tabula.read_pdf(
    arquivo_pdf,
    pages='2-38',
    multiple_tables=True,
    lattice=False,
    stream=True,
    pandas_options={"header": None}
)
```

```python
colunas = ["Ranking 2024", "Ranking 2023", "Razão Social", "Sede", "Faturamento (R$)"]

tabelas_corrigidas = []
for tabela in tabelas:
    if tabela.shape[1] == 5:
        tabela.columns = colunas
        tabelas_corrigidas.append(tabela)

df_final = pd.concat(tabelas_corrigidas, ignore_index=True)
```

### Resolvendo alguns problemas

```python
# Garante que são strings
df[['Ranking 24', 'Ranking 23']] = df['Ranking 2024'].str.split(' ', expand=True, n=1)

df
```

```python
# Preenche valores NaN em 'Ranking 24' com os valores de 'Ranking 2024'
df['Ranking 24'] = df['Ranking 24'].fillna(df['Ranking 2024'])

# Preenche valores NaN em 'Ranking 23' com os valores de 'Ranking 2023'
df['Ranking 23'] = df['Ranking 23'].fillna(df['Ranking 2023'])

# Preenche valores NaN em 'Razão Social' com os valores de 'Ranking 2023'
df['Razão Social'] = df['Razão Social'].fillna(df['Ranking 2023'])

df
```

```python
df.iloc[100:120]
```

```python
df = df.drop(columns=['Ranking 2024', 'Ranking 2023'])
```

```python
# Renomeando as colunas
df = df.rename(columns={
    'Ranking 24': 'Ranking 2024',
    'Ranking 23': 'Ranking 2023',
    'Razão Social': 'Nome da Empresa',
    'Sede': 'Sede',
    'Faturamento (R$)': 'Faturamento Total (R$)'
})

# Reordenando as colunas (coloque na ordem que desejar)
df = df[['Ranking 2024', 'Ranking 2023', 'Nome da Empresa', 'Sede', 'Faturamento Total (R$)']]
```

```python
# Remove as linhas que contêm a palavra "Total" em qualquer coluna
df = df[~df.apply(lambda row: row.astype(str).str.contains('Total', case=False).any(), axis=1)]
df = df[~df.apply(lambda row: row.astype(str).str.contains('2024', case=False).any(), axis=1)]
```
